# DECISION TREE (REGRESSION)

```python
import pandas as pd
df=pd.read_csv(r"C:\Users\Admin\Desktop\student_performance.csv")
df
```

```python
# UNIQUE VALUES IN RESULT
df['result'].unique()
```

```python
# TRANSFORM CATEGORICAL DATA INTO NUMERIC
from sklearn.preprocessing import LabelEncoder
l=LabelEncoder()
```

```python
df['result']=l.fit_transform(df['result'])
```

```python
# DATA TRANSFORMED (PASS=1, FAIL=0)
df['result'].unique()
```

```python
# DECISION TREE (REGRESSION)
# STEP 1: ASSIGNING VARIABLES(INDEPENDENT/DEPENDENT)

# INDEPENDENT VARIABLE
x=df[['weekly_self_study_hours','attendance_percentage','class_participation','total_marks']]

# DEPENDENT VARIABLE
y=df[['result']]
```

```python
# STEP 2:IMPORT TRAIN TEST SPLIT
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3)
```

```python
print("total data: ",df.shape)
print("x_train data: ",x_train.shape)
print("x_test data: ",x_test.shape)
print("y_train data: ",y_train.shape)
print("y_test data: ",y_test.shape)
```

```python
# STEP 3: APPLY DECISION TREE (REGRESSION)
from sklearn.tree import DecisionTreeRegressor 
regressor=DecisionTreeRegressor()
regressor
```

```python
# STEP 4 :FIT TO DATA
# Use only train(x_train,y_train)
regressor.fit(x_train,y_train)
```

```python
# CHECK ACCURACY
regressor.score(x_test,y_test)
```

```python
# PREDICTION OF NEW DATA
# IMPORT NUMPY FIRST(DATA MUST IN ARRAY FORM)
import numpy as np
data=np.array([[11.7,100.0,8.2,69.8]])
data
```

```python
prediction=regressor.predict(data)
prediction

if prediction == 0:
    print("Fail")
elif prediction ==1:
    print("Pass")
else :
    print("Error")
```
