   #                                               LOGISTIC REGRESSION

```python
import pandas as pd
data=pd.read_csv("D:\Machine_Learning\Iris.csv")
data
```

```python
# DATA VISUALIZATION
import seaborn as sn
```

```python
sn.scatterplot(x="SepalLengthCm",y="PetalLengthCm",data=data,hue="Species")
```

```python
# LOGISTIC REGRESSION
# STEP 1: ASSIGNING VARIABLES(INDEPENDENT/DEPENDENT)

# INDEPENDENT VARIABLE
x=data[["SepalLengthCm","SepalWidthCm","PetalLengthCm","PetalWidthCm"]]

# DEPENDENT VARIABLE
y=data[["Species"]]
```

```python
# STEP 2:IMPORT TRAIN TEST SPLIT
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3)
```

```python
print("total data: ",data.shape)
print("x_train data: ",x_train.shape)
print("x_test data: ",x_test.shape)
print("y_train data: ",y_train.shape)
print("y_test data: ",y_test.shape)
```

```python
# STEP 3: APPLY LOGISTIC REGRESSION
from sklearn.linear_model import LogisticRegression 
model=LogisticRegression()
model
```

```python
# STEP 4 :FIT TO DATA
# Use only train(x_train,y_train)
model.fit(x_train,y_train)
```

```python
# CHECK ACCURACY
model.score(x_test,y_test)
```

```python
# PREDICTION OF NEW DATA
# IMPORT NUMPY FIRST(DATA MUST IN ARRAY FORM)
import numpy as np
data=np.array([[6.4,3.0,4.5,2.1]])
data
```

```python
prediction=model.predict(data)
prediction
```
