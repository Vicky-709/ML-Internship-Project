# LINEAR REGRESSION

```python
import pandas as pd
df=pd.read_csv(r"C:\Users\Admin\Desktop\student_info.csv")
df
```

```python
# DATA VISUALIZATION
import seaborn as sn
```

```python
sn.scatterplot(x="study_hours",y="student_marks",data=df)
```

```python
# LINEAR REGRESSION
# STEP 1: ASSIGNING VARIABLES(INDEPENDENT/DEPENDENT)

# INDEPENDENT VARIABLE
x=df[["study_hours"]]

# DEPENDENT VARIABLE
y=df[["student_marks"]]
```

```python
# STEP 2:IMPORT TRAIN TEST SPLIT
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3)
```

```python
print("total data: ",df.shape)
print("x_train data: ",x_train.shape)
print("x_test data: ",x_test.shape)
print("y_train data: ",y_train.shape)
print("y_test data: ",y_test.shape)
```

```python
# STEP 3: APPLY LINEAR REGRESSION
from sklearn.linear_model import LinearRegression 
model=LinearRegression()
model
```

```python
# STEP 4 :FIT TO DATA
# Use only train(x_train,y_train)
model.fit(x_train,y_train)
```

```python
# CHECK ACCURACY
model.score(x_test,y_test)
```

```python
# PREDICTION OF NEW DATA
# IMPORT NUMPY FIRST(DATA MUST IN ARRAY FORM)
import numpy as np
data=np.array([[5]])
data
```

```python
prediction=model.predict(data)
prediction
```
